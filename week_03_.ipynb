```python
import sys
!{sys.executable} -m pip install qiskit

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# X Gate
qc = QuantumCircuit(1)
qc.x(0)
print("X Gate")
print(qc.draw())
print(Statevector.from_instruction(qc))

# Y Gate
qc = QuantumCircuit(1)
qc.y(0)
print("\nY Gate")
print(qc.draw())
print(Statevector.from_instruction(qc))

# Z Gate
qc = QuantumCircuit(1)
qc.h(0)      # Prepare |+>
qc.z(0)
print("\nZ Gate")
print(qc.draw())
print(Statevector.from_instruction(qc))
```

```python
import random
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

gates = ['x','y','z','h','s','t']

qc = QuantumCircuit(1)

sequence = random.choices(gates, k=5)

print("Random Gate Sequence:", sequence)

for gate in sequence:
    if gate == 'x':
        qc.x(0)
    elif gate == 'y':
        qc.y(0)
    elif gate == 'z':
        qc.z(0)
    elif gate == 'h':
        qc.h(0)
    elif gate == 's':
        qc.s(0)
    elif gate == 't':
        qc.t(0)

print(qc.draw())

state = Statevector.from_instruction(qc)

print("Final Statevector:")
print(state)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

# Hadamard creates equal probability
qc.h(0)

print(qc.draw())

state = Statevector.from_instruction(qc)

prob = state.probabilities()

print("Probability of Heads (|0>):", prob[0])
print("Probability of Tails (|1>):", prob[1])

# Simulate one coin flip
result = "Heads" if prob[0] > 0.49 else "Tails"

print("Possible Outcome:", result)
```
