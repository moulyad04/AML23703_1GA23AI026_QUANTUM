# Measure a Hadamard-superposition qubit with 100, 1000 and 10000 shots and tabulate how the measured probability converges to 0.5

```python
import sys
!{sys.executable} -m pip install qiskit qiskit-aer

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

sim = AerSimulator()

for shots in [100, 1000, 10000]:
    qc = QuantumCircuit(1, 1)
    qc.h(0)
    qc.measure(0, 0)

    result = sim.run(transpile(qc, sim), shots=shots).result()
    counts = result.get_counts()
    p0 = counts.get("0", 0) / shots
    p1 = counts.get("1", 0) / shots

    print(shots, counts, p0, p1)
```

# Medium: Measure a qubit in the X-basis (apply H before measurement) after preparing |+> and |-> states and interpret the results

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

sim = AerSimulator()

def measure_x(state):
    qc = QuantumCircuit(1, 1)

    if state == "+":
        qc.h(0)          # |0> -> |+>
    else:
        qc.x(0)
        qc.h(0)          # |1> -> |->

    qc.h(0)              # X-basis measurement
    qc.measure(0, 0)

    return sim.run(qc, shots=1000).result().get_counts()

print("+ state:", measure_x("+"))
print("- state:", measure_x("-"))
```

# Use repeated measurement statistics to estimate an unknown rotation angle applied to a qubit (simple quantum state tomography)

```python
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

theta = np.pi / 3
shots = 10_000

qc = QuantumCircuit(1, 1)
qc.ry(theta, 0)
qc.measure(0, 0)

sim = AerSimulator()
result = sim.run(transpile(qc, sim), shots=shots).result()
counts = result.get_counts()

p1 = counts.get("1", 0) / shots
theta_est = 2 * np.arcsin(np.sqrt(p1))

print("Counts:", counts)
print("Estimated P(1):", p1)
print("Estimated angle:", theta_est)
print("Degrees:", np.degrees(theta_est))
```
